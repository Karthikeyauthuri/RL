```python

import gymnasium as gym
import numpy as np
```

```python

def run_random_agent(env_name, episodes=5):
    env = gym.make(env_name, render_mode=None)
    print(f"\nRunning Random Agent in {env_name} environment")
    rewards = []

    for episode in range(episodes):
        state, info = env.reset()
        done = False
        total_reward = 0

        while not done:
            action = env.action_space.sample()
            next_state, reward, terminated, truncated, info = env.step(action)
            total_reward += reward
            state = next_state
            done = terminated or truncated

        rewards.append(total_reward)
        print(f"Episode {episode+1}: Total Reward = {total_reward}")

    avg_reward = np.mean(rewards)
    print(f"\nAverage Reward over {episodes} episodes in {env_name}: {avg_reward:.2f}")
    env.close()
```

```python
run_random_agent("CliffWalking-v1", episodes=5)
```

```python

run_random_agent("Taxi-v4", episodes=5)
```
