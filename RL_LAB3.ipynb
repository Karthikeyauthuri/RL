```python

import random

machines = ['A', 'B', 'C']
epsilon = 0.2
plays = 2

N = {'A': 0, 'B': 0, 'C': 0}
R = {'A': [], 'B': [], 'C': []}
Q = {'A': 0, 'B': 0, 'C': 0}

total_reward = 0

for i in range(plays):

    r = random.random()
    highest = max(Q, key=Q.get)

    if r < epsilon:
        choices = [m for m in machines if m != highest]
        machine = random.choice(choices)
        method = "Exploration"
    else:
        machine = highest
        method = "Exploitation"

    print("\nPlay", i + 1)
    print("Random value:", round(r, 3))
    print("Method:", method)
    print("Selected Machine:", machine)

    reward = float(input("Enter reward: "))

    N[machine] += 1
    R[machine].append(reward)

    Q[machine] = Q[machine] + \
        (reward - Q[machine]) / N[machine]

    total_reward += reward

    print("N =", N[machine])
    print("Q =", round(Q[machine], 2))

print("\n===== FINAL RESULT =====")

for machine in machines:
    print(machine, "N =", N[machine],
          "Rewards =", R[machine],
          "Q =", round(Q[machine], 2))

print("Total Reward =", total_reward)
print("Best Machine =", max(Q, key=Q.get))
```
